```python
import SimpleITK as sitk
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt

def show_resize_effect(image_path, resize_size=(128, 256, 256), slice_index=None):
    # Orijinal veriyi oku
    image = sitk.ReadImage(image_path)
    image_np = sitk.GetArrayFromImage(image).astype("float32")  # [D, H, W]

    # Normalize
    image_np = (image_np - image_np.min()) / (image_np.max() - image_np.min() + 1e-8)

    # Tensöre çevir ve resize
    image_tensor = torch.from_numpy(image_np).unsqueeze(0).unsqueeze(0)  # [1, 1, D, H, W]
    resized = F.interpolate(image_tensor, size=resize_size, mode="trilinear", align_corners=False)
    resized_np = resized.squeeze().cpu().numpy()  # [D, H, W]

    # Slice seçimi
    if slice_index is None:
        slice_index = image_np.shape[0] // 2

    fig, axs = plt.subplots(1, 2, figsize=(12, 6))

    axs[0].imshow(image_np[slice_index], cmap="gray")
    axs[0].set_title(f"Orijinal - {image_np.shape}")

    scaled_z = int(slice_index * resize_size[0] / image_np.shape[0])  # Z ekseniyle orantıla
    axs[1].imshow(resized_np[scaled_z], cmap="gray")
    axs[1].set_title(f"Resize - {resize_size}")

    for ax in axs:
        ax.axis("off")

    plt.tight_layout()
    plt.show()
```

```python
import SimpleITK as sitk
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
import numpy as np
def show_resize_effect_on_mask(mask_path, resize_size=(128, 256, 256), slice_index=None):
    # Maskeyi yükle (etiketli veri)
    mask = sitk.ReadImage(mask_path)
    mask_np = sitk.GetArrayFromImage(mask).astype("int64")  # [D, H, W]

    # Tensöre çevir ve resize
    mask_tensor = torch.from_numpy(mask_np).unsqueeze(0).unsqueeze(0).float()  # [1, 1, D, H, W]
    resized = F.interpolate(mask_tensor, size=resize_size, mode="nearest")
    resized_np = resized.squeeze().cpu().numpy().astype("int64")  # [D, H, W]

    # Slice seçimi
    if slice_index is None:
        slice_index = mask_np.shape[0] // 2
    scaled_z = int(slice_index * resize_size[0] / mask_np.shape[0])

    # Sınıf istatistiklerini yazdır
    unique_before = np.unique(mask_np)
    unique_after = np.unique(resized_np)
    print(f"🎯 Orijinal maske sınıfları: {unique_before}")
    print(f"🔁 Resize sonrası sınıflar:  {unique_after}")

    # Görselleştir
    fig, axs = plt.subplots(1, 2, figsize=(12, 6))

    axs[0].imshow(mask_np[slice_index], cmap="nipy_spectral")
    axs[0].set_title(f"Orijinal Maske - {mask_np.shape}")

    axs[1].imshow(resized_np[scaled_z], cmap="nipy_spectral")
    axs[1].set_title(f"Resize Maske - {resize_size}")

    for ax in axs:
        ax.axis("off")

    plt.tight_layout()
    plt.show()
```
